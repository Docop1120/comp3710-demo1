```python
import torch
import numpy as np
```

```python
print(torch.version.cuda)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)
```

```python
# grid for computing image , subdivide the space
X , Y = np . mgrid [ -4.0:4:0.01 , -4.0:4:0.01]
```

```python
# load into PyTorch tensors
x = torch . Tensor ( X )
y = torch . Tensor ( Y )
```

```python
# transfer to the GPU device
x = x . to ( device )
y = y . to ( device )
```

```python
# plot
import matplotlib . pyplot as plt
z = torch . exp ( -( x **2+ y **2) /2.0)
plt . imshow ( z . cpu () . numpy () ) # Updated !
plt . tight_layout ()
plt . show ()
```

```python
# converting Gaussian functing into 2d sine
v = torch.sin (2 * np . pi* (x + y))
plt . imshow ( v . cpu () . numpy () ) # Updated !
plt . tight_layout ()
plt . show ()
```

```python
# combine sine and Gaussian
w = v * z
plt . imshow ( w . cpu () . numpy () ) # Updated !
plt . tight_layout ()
plt . show ()
```
